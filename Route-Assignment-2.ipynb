```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = {}

    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price

    def book_tables(self, customer, table_no):
        self.book_table.append((customer, table_no))

    def customer_order(self, customer, item):
        if item in self.menu_items:
            if customer not in self.customer_orders:
                self.customer_orders[customer] = []
            self.customer_orders[customer].append(item)
        else:
            print("Sorry,", item, "is not on the menu")
```

```python
r = Restaurant()

r.add_item_to_menu("Burger", 5)
r.add_item_to_menu("Pizza", 8)
r.add_item_to_menu("Juice", 2)

r.book_tables("Ali", 1)
r.book_tables("Sara", 2)

r.customer_order("Ali", "Burger")
r.customer_order("Ali", "Juice")
r.customer_order("Sara", "Pizza")

print("Menu:", r.menu_items)
print("Table reservations:", r.book_table)
print("Customer orders:", r.customer_orders)
```

```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_age(self):
        return self.age

    def set_age(self, age):
        self.age = age
```

```python
class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, name):
        self.name = name

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price
```

```python
class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes):
        self.bikes = bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, rentals):
        self.rentals = rentals

    def addBike(self, bike):
        self.bikes.append(bike)

    def requestBike(self, customer, num_of_bikes):
        if customer.get_age() <= 6:
            print("Customer must be older than 6 years")
        elif num_of_bikes > len(self.bikes):
            print("Wait 10 mins")
        else:
            for i in range(num_of_bikes):
                self.rentals.append(self.bikes.pop())
            print(customer.get_name(), "rented", num_of_bikes, "bike(s)")

    def returnBike(self, rentalTime, num_of_bikes):
        for i in range(num_of_bikes):
            self.bikes.append(self.rentals.pop())
        bill = rentalTime * 40
        print("Total bill:", bill)
        return bill

    def totalCost(self):
        total = 0
        for b in self.bikes:
            total += b.get_price()
        return total
```

```python
shop = MainBikeRental()
shop.addBike(Bike("BMX", 100))
shop.addBike(Bike("Mountain", 150))
shop.addBike(Bike("Road", 200))

c = Customer("Ali", 20)
kid = Customer("Sam", 5)

shop.requestBike(kid, 1)
shop.requestBike(c, 5)
shop.requestBike(c, 2)
print("Bikes left:", len(shop.get_bikes()))
shop.returnBike(3, 2)
print("Bikes in stock:", len(shop.get_bikes()))
print("Total cost of bikes in store:", shop.totalCost())
```
